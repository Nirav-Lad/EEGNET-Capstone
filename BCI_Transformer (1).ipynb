```python
import numpy as np
import scipy.io
from google.colab import drive

# --- Mount Drive and Load File (if needed) ---
# drive.mount('/content/drive')
# file_path = '/content/drive/My Drive/Research-BCI/Research-BCI/archive (1)/A01T.mat'

# --- The Definitive Data Extraction ---

# 1. Load the data using the squeeze_me=True argument
mat_data_squeezed = scipy.io.loadmat(file_path, squeeze_me=True)

# 2. Access the array containing all subjects' data
all_subjects_data = mat_data_squeezed['data']

# 3. Select the data for a subject that has labels (e.g., the 4th subject at index 3)
subject_data = all_subjects_data[3] # Using Subject 4

# 4. Extract the scalar object for each field, then use .item() to get the array inside
X = subject_data['X'].item()
trial_starts = subject_data['trial'].item()
y = subject_data['y'].item()
sampling_freq = subject_data['fs'].item()
class_names = subject_data['classes'].item()

# --- Final Verification ---
print("--- Data Extraction Verified ---")
print(f"Shape of X (EEG signals): {X.shape}")
print(f"Shape of trial_starts (trial start samples): {trial_starts.shape}")
print(f"Shape of y (class labels): {y.shape}")
print(f"Number of trials: {len(y)}")
print(f"Sampling Frequency: {sampling_freq} Hz")
print(f"Class Names: {class_names}")
print(f"Unique Labels in y: {np.unique(y)}")
```

```python
# --- Epoching Parameters ---
# We define the time window for each epoch in seconds
# Let's take a 4-second window starting from the stimulus onset
epoch_start_s = 0.0  # 0 seconds after the marker
epoch_end_s = 4.0    # 4 seconds after the marker

# Convert the time window into sample points
epoch_start_samples = int(epoch_start_s * sampling_freq)
epoch_end_samples = int(epoch_end_s * sampling_freq)
epoch_length_samples = epoch_end_samples - epoch_start_samples

# --- Create the Epochs ---
# We will store our epochs in a list
epochs = []

# Loop through each trial start time
for start_sample in trial_starts:
    # Define the start and end point of the slice in the X array
    start_index = start_sample + epoch_start_samples
    end_index = start_sample + epoch_end_samples

    # Slice the X array to get one epoch
    single_epoch = X[start_index:end_index, :]

    # Append the epoch to our list
    epochs.append(single_epoch)

# Convert the list of epochs into a single 3D NumPy array
# The shape will be (number_of_trials, samples_per_epoch, number_of_channels)
epochs_array = np.array(epochs)


# --- Verification ---
print("--- Epoching Complete ---")
print(f"Shape of the final epoched data (trials, samples, channels): {epochs_array.shape}")
print(f"Number of trials should match y: {len(epochs_array)}")
print(f"Number of samples per trial should be {epoch_length_samples}: {epochs_array.shape[1]}")
print(f"Number of channels should be 25: {epochs_array.shape[2]}")
```

```python
import tensorflow as tf
from tensorflow.keras.layers import Input, Dense, LayerNormalization, Dropout, MultiHeadAttention, GlobalAveragePooling1D
from tensorflow.keras.models import Model

# --- Transformer Encoder Block ---
def transformer_encoder(inputs, head_size, num_heads, ff_dim, dropout=0):
    # Attention and Normalization
    x = MultiHeadAttention(key_dim=head_size, num_heads=num_heads, dropout=dropout)(inputs, inputs)
    x = Dropout(dropout)(x)
    x = LayerNormalization(epsilon=1e-6)(x + inputs)

    # Feed Forward Part
    ffn = tf.keras.Sequential([
        Dense(ff_dim, activation="relu"),
        Dense(inputs.shape[-1]),
    ])
    x = ffn(x)
    x = Dropout(dropout)(x)
    x = LayerNormalization(epsilon=1e-6)(x)
    return x

# --- Build the Full Model ---
def build_transformer_model(input_shape, head_size, num_heads, ff_dim, num_transformer_blocks, num_classes, dropout=0):
    inputs = Input(shape=input_shape)
    x = inputs

    # Create multiple Transformer Encoder blocks
    for _ in range(num_transformer_blocks):
        x = transformer_encoder(x, head_size, num_heads, ff_dim, dropout)

    # Global Average Pooling to reduce the sequence to a single feature vector
    x = GlobalAveragePooling1D(data_format="channels_last")(x)

    # Final Classification Head
    x = Dropout(0.4)(x)
    x = Dense(20, activation="relu")(x)
    outputs = Dense(num_classes, activation="softmax")(x)

    return Model(inputs, outputs)

# --- Define Model Hyperparameters ---
INPUT_SHAPE = X_train_processed.shape[1:]  # (1000, 25)
NUM_CLASSES = len(np.unique(y))       # 4 classes

# Transformer-specific hyperparameters
HEAD_SIZE = 256
NUM_HEADS = 4
FF_DIM = 4  # Feed-forward network dimension
NUM_TRANSFORMER_BLOCKS = 4
DROPOUT_RATE = 0.25


# --- Create the model instance ---
model = build_transformer_model(
    INPUT_SHAPE,
    head_size=HEAD_SIZE,
    num_heads=NUM_HEADS,
    ff_dim=FF_DIM,
    num_transformer_blocks=NUM_TRANSFORMER_BLOCKS,
    num_classes=NUM_CLASSES,
    dropout=DROPOUT_RATE,
)

# Compile the model
model.compile(
    loss="sparse_categorical_crossentropy",
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-4),
    metrics=["sparse_categorical_accuracy"],
)

# --- Verification ---
# Print a summary of the model architecture
print("--- Model Architecture Summary ---")
model.summary()
```

```python
import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense, Activation, Dropout, Flatten
from tensorflow.keras.layers import Conv2D, AveragePooling2D, BatchNormalization, DepthwiseConv2D, SeparableConv2D
from tensorflow.keras.callbacks import EarlyStopping

# --- 1. Reshape Data for EEGNet (Important!) ---
# EEGNet uses 2D convolutions, so it expects the data in a 4D format:
# (trials, channels, samples, 1)
# Our data is currently (trials, samples, channels), so we need to swap the axes and add a dimension.

X_train_eegnet = X_train_processed.transpose(0, 2, 1)
X_test_eegnet = X_test_processed.transpose(0, 2, 1)

X_train_eegnet = X_train_eegnet[..., tf.newaxis]
X_test_eegnet = X_test_eegnet[..., tf.newaxis]

print(f"Reshaped X_train for EEGNet: {X_train_eegnet.shape}")
print(f"Reshaped X_test for EEGNet: {X_test_eegnet.shape}")


# --- 2. Build the EEGNet Model ---
def EEGNet(nb_classes, Chans=25, Samples=1000, dropoutRate=0.5, kernLength=64, F1=8, D=2, F2=16):
    input1 = Input(shape=(Chans, Samples, 1))

    # Block 1
    block1 = Conv2D(F1, (1, kernLength), padding='same', use_bias=False)(input1)
    block1 = BatchNormalization()(block1)
    block1 = DepthwiseConv2D((Chans, 1), use_bias=False, depth_multiplier=D, depthwise_constraint=tf.keras.constraints.max_norm(1.))(block1)
    block1 = BatchNormalization()(block1)
    block1 = Activation('elu')(block1)
    block1 = AveragePooling2D((1, 4))(block1)
    block1 = Dropout(dropoutRate)(block1)

    # Block 2
    block2 = SeparableConv2D(F2, (1, 16), use_bias=False, padding='same')(block1)
    block2 = BatchNormalization()(block2)
    block2 = Activation('elu')(block2)
    block2 = AveragePooling2D((1, 8))(block2)
    block2 = Dropout(dropoutRate)(block2)

    # Flatten and Classify
    flatten = Flatten(name='flatten')(block2)
    dense = Dense(nb_classes, name='dense', kernel_constraint=tf.keras.constraints.max_norm(0.25))(flatten)
    softmax = Activation('softmax', name='softmax')(dense)

    return Model(inputs=input1, outputs=softmax)

# --- 3. Create and Compile the Model ---
model_eegnet = EEGNet(nb_classes=NUM_CLASSES, Chans=25, Samples=1000)

model_eegnet.compile(loss='sparse_categorical_crossentropy', optimizer='adam', metrics=['sparse_categorical_accuracy'])
model_eegnet.summary()

# --- 4. Train the EEGNet Model ---
# Labels are already adjusted from the previous step (y_train_adjusted, y_test_adjusted)
early_stopping = EarlyStopping(monitor='val_loss', patience=25, restore_best_weights=True)

print("\n--- Starting EEGNet Model Training ---")
history_eegnet = model_eegnet.fit(
    X_train_eegnet,
    y_train_adjusted,
    batch_size=8,
    epochs=150,
    validation_data=(X_test_eegnet, y_test_adjusted),
    callbacks=[early_stopping],
    verbose=1
)
print("\n--- EEGNet Model Training Complete! ---")
```

```python
import tensorflow as tf
from tensorflow.keras.layers import Input, Dense, LayerNormalization, Dropout, MultiHeadAttention, GlobalAveragePooling1D
from tensorflow.keras.models import Model
from tensorflow.keras.callbacks import EarlyStopping
import numpy as np

# Note: This code assumes you have the variables from the previous step:
# X_train_processed, y_train_adjusted, X_test_processed, y_test_adjusted

# --- 1. Build the Transformer Model (using the simplified version from our last attempt) ---
def transformer_encoder(inputs, head_size, num_heads, ff_dim, dropout=0):
    x = MultiHeadAttention(key_dim=head_size, num_heads=num_heads, dropout=dropout)(inputs, inputs)
    x = Dropout(dropout)(x)
    x = LayerNormalization(epsilon=1e-6)(x + inputs)
    ffn = tf.keras.Sequential([Dense(ff_dim, activation="relu"), Dense(inputs.shape[-1]),])
    x = ffn(x)
    x = Dropout(dropout)(x)
    x = LayerNormalization(epsilon=1e-6)(x)
    return x

def build_transformer_model(input_shape, head_size, num_heads, ff_dim, num_transformer_blocks, num_classes, dropout=0):
    inputs = Input(shape=input_shape)
    x = inputs
    for _ in range(num_transformer_blocks):
        x = transformer_encoder(x, head_size, num_heads, ff_dim, dropout)
    x = GlobalAveragePooling1D(data_format="channels_last")(x)
    x = Dropout(0.4)(x)
    x = Dense(20, activation="relu")(x)
    outputs = Dense(num_classes, activation="softmax")(x)
    return Model(inputs, outputs)

# --- 2. Define Model Hyperparameters ---
INPUT_SHAPE = X_train_processed.shape[1:]
NUM_CLASSES = len(np.unique(y))
NUM_TRANSFORMER_BLOCKS = 2
HEAD_SIZE = 128
NUM_HEADS = 4
FF_DIM = 4
DROPOUT_RATE = 0.25

# --- 3. Create and Compile the Model ---
model_transformer = build_transformer_model(
    INPUT_SHAPE,
    head_size=HEAD_SIZE,
    num_heads=NUM_HEADS,
    ff_dim=FF_DIM,
    num_transformer_blocks=NUM_TRANSFORMER_BLOCKS,
    num_classes=NUM_CLASSES,
    dropout=DROPOUT_RATE,
)

optimizer = tf.keras.optimizers.Adam(learning_rate=1e-3)
model_transformer.compile(
    loss="sparse_categorical_crossentropy",
    optimizer=optimizer,
    metrics=["sparse_categorical_accuracy"],
)
model_transformer.summary()


# --- 4. Train the Transformer on the Full Dataset ---
early_stopping = EarlyStopping(monitor='val_loss', patience=15, restore_best_weights=True)

print("\n--- Starting Transformer Model Training on Full Dataset ---")
history_transformer = model_transformer.fit(
    X_train_processed,
    y_train_adjusted,
    batch_size=16,
    epochs=150,
    validation_data=(X_test_processed, y_test_adjusted),
    callbacks=[early_stopping],
    verbose=1
)
print("\n--- Transformer Model Training Complete! ---")


# --- 5. Final Evaluation ---
print("\n--- Final Transformer Model Evaluation ---")
final_loss_transformer, final_accuracy_transformer = model_transformer.evaluate(X_test_processed, y_test_adjusted, verbose=0)
print(f"Final Test Loss: {final_loss_transformer:.4f}")
print(f"Final Test Accuracy: {final_accuracy_transformer:.4f} ({(final_accuracy_transformer * 100):.2f}%)")
```

```python
import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense, Dropout, BatchNormalization
from tensorflow.keras.layers import Conv1D, MaxPooling1D, LSTM
from tensorflow.keras.callbacks import EarlyStopping

# Note: This code assumes you have the variables from the aggregation step:
# X_train_processed, y_train_adjusted, X_test_processed, y_test_adjusted

print("--- Data Shapes for CNN-LSTM ---")
print(f"X_train_processed shape: {X_train_processed.shape}")
print(f"X_test_processed shape: {X_test_processed.shape}")


# --- 1. Build the CNN-LSTM Model ---
def build_cnnlstm_model(input_shape, num_classes):
    inputs = Input(shape=input_shape)

    # CNN Block
    x = Conv1D(filters=64, kernel_size=16, activation='relu')(inputs)
    x = BatchNormalization()(x)
    x = MaxPooling1D(pool_size=2)(x)
    x = Dropout(0.5)(x)

    x = Conv1D(filters=64, kernel_size=16, activation='relu')(x)
    x = BatchNormalization()(x)
    x = MaxPooling1D(pool_size=2)(x)
    x = Dropout(0.5)(x)

    # LSTM Block
    x = LSTM(128, return_sequences=False)(x) # return_sequences=False because it's the last LSTM
    x = Dropout(0.5)(x)

    # Classification Head
    outputs = Dense(num_classes, activation='softmax')(x)

    return Model(inputs, outputs)

# --- 2. Create and Compile the Model ---
INPUT_SHAPE = X_train_processed.shape[1:] # (1000, 25)
NUM_CLASSES = len(np.unique(y))       # 4 classes

model_cnnlstm = build_cnnlstm_model(INPUT_SHAPE, NUM_CLASSES)
model_cnnlstm.compile(loss='sparse_categorical_crossentropy', optimizer='adam', metrics=['sparse_categorical_accuracy'])
model_cnnlstm.summary()


# --- 3. Train the CNN-LSTM Model ---
early_stopping = EarlyStopping(monitor='val_loss', patience=25, restore_best_weights=True)

print("\n--- Starting CNN-LSTM Model Training ---")
history_cnnlstm = model_cnnlstm.fit(
    X_train_processed,
    y_train_adjusted,
    batch_size=16,
    epochs=150,
    validation_data=(X_test_processed, y_test_adjusted),
    callbacks=[early_stopping],
    verbose=1
)
print("\n--- CNN-LSTM Model Training Complete! ---")


# --- 4. Final Evaluation ---
print("\n--- Final CNN-LSTM Model Evaluation ---")
final_loss_cnnlstm, final_accuracy_cnnlstm = model_cnnlstm.evaluate(X_test_processed, y_test_adjusted, verbose=0)
print(f"Final Test Loss: {final_loss_cnnlstm:.4f}")
print(f"Final Test Accuracy: {final_accuracy_cnnlstm:.4f} ({(final_accuracy_cnnlstm * 100):.2f}%)")
```
